# Part 3. Data Collection

## 3.1 Data Requirements

To investigate the macroeconomic determinants of EUR/USD exchange rate movements, a comprehensive dataset must be assembled combining:

1. Macroeconomic indicators for the United States and the Euro Area;
2. Financial market variables, represented by the short-term sovereign yield spread;
3. Foreign exchange market data, represented by the EUR/USD exchange rate.

The dataset covers the period from **1 March 2016 to 1 March 2026**, providing a sufficiently long sample to encompass several distinct macroeconomic regimes, including:

- The pre-COVID expansion period (2016--2019);
- The COVID-19 shock and recovery (2020--2021);
- The global inflation surge and aggressive monetary tightening cycle (2022--2024);
- The subsequent normalization period (2025--2026).

This broad time horizon allows the analysis to capture structural changes in the relationship between macroeconomic fundamentals and exchange rate dynamics.

---

## 3.2 Data Frequency

The variables used in this study are observed at different frequencies:

| Variable Type | Frequency |
|:-------------|:----------|
| EUR/USD exchange rate | Daily |
| 2-year yield spread | Daily |
| CPI/HICP inflation | Monthly |
| Nonfarm Payrolls | Monthly |
| Unemployment rates | Monthly |
| Retail Sales | Monthly |
| PMI indicators | Monthly |

Because exchange rates and bond yields are available on a daily basis while macroeconomic indicators are typically released monthly, the data must later be transformed to a common frequency. The methodology for this harmonization is described in Part 4.

## 3.3 Data Sources

The analysis relies on publicly available and highly reputable data providers.

### 3.3.1 Foreign Exchange Data

The EUR/USD exchange rate series is obtained from:

- European Central Bank (ECB);
- Yahoo Finance;
- Stooq.

The exchange rate is expressed as the number of US dollars per one euro (USD per EUR).

### 3.3.2 Macroeconomic Data

US and Euro Area macroeconomic indicators are sourced from:

- Federal Reserve Economic Data (FRED);
- U.S. Bureau of Labor Statistics (BLS);
- U.S. Census Bureau;
- Institute for Supply Management (ISM);
- Eurostat;
- S\&P Global.

### 3.3.3 Interest Rate Data

The sovereign yield spread is constructed using:

- US 2-Year Treasury yield;
- German 2-Year government bond yield (Schatz).

These series are available through FRED and the Deutsche Bundesbank.

The spread is defined as:

$$
\text{Yield Spread}_t
=
Y^{US}_{2Y,t}
-
Y^{DE}_{2Y,t}
$$

where:

- $Y^{US}_{2Y,t}$ is the US 2-year Treasury yield;
- $Y^{DE}_{2Y,t}$ is the German 2-year government bond yield.

## 3.5 Construction of the Target Variable

The dependent variable is the daily logarithmic return of EUR/USD:

$$
r_t
=
\ln\left(
\frac{EURUSD_t}{EURUSD_{t-1}}
\right)
$$

where:

- $EURUSD_t$ denotes the exchange rate at day $t$;
- $r_t$ represents the continuously compounded daily return.

Log returns are preferred because they are additive over time and generally exhibit more stable statistical properties than raw price levels.

## 3.6 Data Release Timing and Information Set

An important methodological issue concerns the timing of macroeconomic announcements.

Financial markets react not to the observed value itself, but to the information revealed relative to prior expectations. Consequently, each macroeconomic observation becomes economically relevant only on its official release date.

To reflect this information structure:

1. Each indicator is aligned with its publication date.
2. The released value remains in effect until the next publication.
3. Alternative specifications may later consider standardized economic surprises when consensus forecasts are available.

This approach ensures that the explanatory variables represent information that was actually observable by market participants at each point in time.
