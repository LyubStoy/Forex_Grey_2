# Project Structure and Research Plan

This project will be conducted in a series of structured steps designed to progressively identify, analyze, and quantify the macroeconomic determinants of EUR/USD exchange rate fluctuations.

---

# Part 1. Research Objective and Economic Framework

The first part of the study defines the research question and establishes the theoretical framework underlying the analysis.

### Research Question

> Do macroeconomic indicators explain daily EUR/USD fluctuations better than the yield spread alone?

### Economic Transmission Mechanism

$$
\text{Macroeconomic Data}
\rightarrow
\text{Monetary Policy Expectations}
\rightarrow
\text{Yield Spread}
\rightarrow
\text{Capital Flows}
\rightarrow
\text{EUR/USD}
$$

---

# Part 6. Hypothesis Testing

Formal statistical tests will be used to assess the significance of the relationships.

### Null Hypothesis

$$
H_0:
\text{Macroeconomic indicators do not significantly explain EUR/USD fluctuations.}
$$

### Alternative Hypothesis

$$
H_1:
\text{At least one macroeconomic indicator significantly explains EUR/USD fluctuations.}
$$

---

# Part 7. Econometric Modeling

Several regression models will be estimated and compared.

### Model 1: Benchmark Model
- EUR/USD returns explained by the yield spread only.

### Model 2: Extended Macroeconomic Model
- Yield spread plus macroeconomic indicators.

### Model 3: Reduced Model
- Only statistically significant variables retained.

### Evaluation Metrics
- Coefficient signs and magnitudes
- p-values
- Confidence intervals
- Adjusted $R^2$
- Residual diagnostics

---
