# Part 1. Research Objective and Economic Framework

## 1.1 Research Objective

The objective of this project is to investigate whether key macroeconomic indicators from the United States and the Euro Area can explain daily fluctuations in the EUR/USD exchange rate more effectively than the yield spread alone.

Building on a previous empirical analysis that demonstrated that the US–Euro Area yield spread has a statistically significant but economically limited explanatory power, this study aims to identify the fundamental macroeconomic drivers that shape monetary policy expectations and influence exchange rate dynamics.

More specifically, the project examines whether indicators related to inflation, labor market conditions, economic growth, and consumer demand provide additional explanatory power for EUR/USD movements beyond that contained in the yield differential.

The ultimate goal is to develop a data-driven framework capable of:

1. Identifying the most influential macroeconomic indicators for EUR/USD.
2. Quantifying their statistical significance and economic impact.
3. Comparing their explanatory power with that of the yield spread.
4. Laying the foundation for a real-time monitoring tool that tracks macroeconomic releases, financial news, and market sentiment to assess their potential impact on the EUR/USD exchange rate.

---

## 1.2 Economic Motivation

The first stage of the Forex Build Tool project showed that changes in the US–Euro Area yield spread are significantly associated with changes in the EUR/USD exchange rate. However, the explanatory power of this relationship was found to be relatively modest, with the yield differential accounting for only a limited fraction of daily exchange rate variability.

This finding suggests that the yield spread, while informative, should not be interpreted as the fundamental source of exchange rate movements. Rather, it reflects market expectations regarding the future path of monetary policy in the United States and the euro area.

Since these expectations are formed in response to incoming macroeconomic information, a more fundamental approach consists of analyzing the economic indicators that influence central bank decisions and, ultimately, exchange rate dynamics.

---

## 1.3 Economic Transmission Mechanism

The macro-financial transmission process underlying the analysis can be summarized as follows:

$$
\text{Macroeconomic Data}
\rightarrow
\text{Monetary Policy Expectations}
\rightarrow
\text{Yield Spread}
\rightarrow
\text{Capital Flows}
\rightarrow
\text{EUR/USD}
$$

Within this framework:

- Macroeconomic indicators affect expectations about future interest rates.
- Monetary policy expectations are incorporated into sovereign bond yields.
- Yield differentials influence international capital allocation.
- Capital flows affect the EUR/USD exchange rate.

---

## 1.4 Research Question

The central research question addressed in this study is:

> Do macroeconomic indicators explain daily EUR/USD fluctuations better than the yield spread alone?

---

## 1.5 Research Hypothesis

### Null Hypothesis ($H_0$)

Macroeconomic indicators do not provide additional explanatory power for EUR/USD fluctuations beyond that contained in the yield spread.

### Alternative Hypothesis ($H_1$)

Macroeconomic indicators provide statistically significant additional explanatory power for EUR/USD fluctuations beyond that contained in the yield spread.

---

## 1.6 Expected Contribution

This study seeks to contribute to the understanding of exchange rate dynamics by:

1. Identifying the macroeconomic variables most relevant to EUR/USD movements.
2. Quantifying their individual and joint explanatory power.
3. Comparing their impact with that of the yield spread.
4. Providing a methodological foundation for the development of a real-time macroeconomic monitoring system for foreign exchange analysis.
