# Part 2. Selection of Macroeconomic Indicators

In order to achieve the objective of this study, it is first necessary to identify the macroeconomic indicators that are most closely monitored by the Federal Reserve (Fed) and the European Central Bank (ECB). Since monetary policy expectations play a central role in determining sovereign bond yields, capital flows, and exchange rate dynamics, the selected variables should reflect the key information sets used by both central banks when assessing economic conditions and setting policy rates.

Although central banks analyze a wide range of economic and financial data, a relatively small set of indicators consistently receives the greatest attention. These indicators can be grouped into four main categories: inflation, labor market conditions, economic activity, and consumer demand.

---

## 2.1 Inflation Indicators

Price stability is the primary objective of both the Federal Reserve and the European Central Bank. Consequently, inflation measures constitute the most important inputs into the monetary policy decision-making process.

### Federal Reserve

The Federal Reserve places particular emphasis on:

- Core Personal Consumption Expenditures (Core PCE) Price Index
- Consumer Price Index (CPI)
- Core CPI

### European Central Bank

The ECB primarily monitors:

- Harmonised Index of Consumer Prices (HICP)
- Core HICP

---

## 2.2 Labor Market Indicators

Labor market conditions provide valuable information about economic slack, wage pressures, and the potential persistence of inflation.

### Federal Reserve

Given its dual mandate of price stability and maximum employment, the Federal Reserve closely monitors:

- Nonfarm Payrolls (NFP)
- Unemployment Rate
- Average Hourly Earnings
- Job Openings and Labor Turnover Survey (JOLTS)

### European Central Bank

The ECB also pays close attention to labor market developments, including:

- Euro Area Unemployment Rate
- Wage Growth
- Negotiated Wage Indicators

---

## 2.3 Economic Activity Indicators

Measures of economic growth and business sentiment help central banks assess the strength and sustainability of economic expansion.

Key indicators include:

- Gross Domestic Product (GDP) Growth
- Industrial Production
- Purchasing Managers' Indices (PMI)
- ISM Manufacturing and Services Indices (United States)

---

## 2.4 Consumer Demand Indicators

Consumer spending is a major component of aggregate demand and an important source of inflationary pressure.

Relevant indicators include:

- Retail Sales
- Consumer Confidence Surveys

---

## 2.5 Financial Conditions

Central banks also consider broader financial conditions, which influence the transmission of monetary policy to the real economy.

Important variables include:

- Government bond yields
- Credit spreads
- Equity market performance
- Exchange rate developments

---

## 2.6 Summary of the Monetary Policy Reaction Function

The policy reaction function of the Federal Reserve and the European Central Bank can be summarized as follows:

$$
\text{Inflation}
+
\text{Labor Market Conditions}
+
\text{Economic Activity}
+
\text{Consumer Demand}
+
\text{Financial Conditions}
\rightarrow
\text{Monetary Policy Decision}
$$

Monetary policy decisions and expectations are subsequently reflected in sovereign bond yields, international capital flows, and exchange rate movements.

## 2.7 Indicators Selected for the Empirical Analysis

Based on their importance in the monetary policy process and their availability over a sufficiently long historical period, the following indicators are selected for the empirical analysis.

| Indicator | Region | Economic Meaning | Why Central Banks Monitor It | Expected Impact on EUR/USD |
|----------|--------|----------------|-----------------------------|----------------------------|
| Consumer Price Index (CPI) | US / Euro Area | Measures changes in consumer prices and overall inflation. | Inflation is the primary objective of monetary policy. | Higher US CPI tends to strengthen the USD and lower EUR/USD; higher Euro Area CPI tends to strengthen the EUR and raise EUR/USD. |
| Nonfarm Payrolls (NFP) | United States | Measures the monthly change in non-agricultural employment. | Provides a timely signal of labor market strength. | Stronger-than-expected NFP generally strengthens the USD and lowers EUR/USD. |
| Unemployment Rate | US / Euro Area | Percentage of the labor force actively seeking work. | Indicates labor market slack and wage pressure. | Lower unemployment generally supports the domestic currency. |
| Retail Sales | United States | Measures consumer spending. | Consumption is the largest component of GDP. | Strong retail sales tend to strengthen the USD and lower EUR/USD. |
| ISM Manufacturing PMI | United States | Survey-based indicator of manufacturing activity. | Provides an early signal of business cycle conditions. | Higher PMI readings usually support the USD. |
| Fed Funds Rate | United States | Main policy interest rate of the Federal Reserve. | Direct instrument used to control inflation. | Higher rates generally strengthen the USD. |
| ECB Deposit Facility Rate | Euro Area | Main policy rate of the ECB. | Core instrument used to steer monetary conditions. | Higher ECB rates generally strengthen the EUR. |
| Euro Area HICP | Euro Area | Harmonised Index of Consumer Prices. | Official inflation benchmark used by the ECB. | Higher HICP tends to strengthen the EUR. |
| 2-Year Yield Spread (US--Germany) | Relative | Difference between US and German short-term yields. | Captures relative monetary policy expectations. | A wider US yield advantage tends to strengthen the USD and lower EUR/USD. |
| Average Hourly Earnings | United States | Measures wage growth. | Wage growth can generate inflationary pressure. | Strong wage growth tends to strengthen the USD. |
| Consumer Confidence | US / Euro Area | Measures household sentiment. | Signals future consumption and growth. | Higher confidence generally supports the domestic currency. |

### United States

- Consumer Price Index (CPI)
- Nonfarm Payrolls (NFP)
- Unemployment Rate
- Retail Sales
- ISM Manufacturing PMI

### Euro Area

- Harmonised Index of Consumer Prices (HICP)
- Unemployment Rate
- Manufacturing PMI

### Financial Variable

- US 2-Year Treasury Yield minus German 2-Year Government Bond Yield

### Target Variable

- Daily EUR/USD returns

---

## 2.8 Rationale for the Selection

The selected variables capture the principal macroeconomic forces that influence central bank decisions and shape market expectations regarding future interest rate differentials. By combining inflation, labor market, growth, and consumption indicators from both economies with a market-based measure of monetary policy expectations, this dataset provides a comprehensive framework for analyzing the fundamental drivers of EUR/USD exchange rate fluctuations.

In particular, the inclusion of the US--German 2-year yield spread allows the analysis to incorporate market expectations directly, while the macroeconomic indicators provide insight into the fundamental information that drives those expectations. Together, these variables offer a robust basis for both econometric modeling and machine learning applications.
